**1.Implementación de la clase**

```python
class Libro:

    # Constructor
    def __init__(self, titulo, autor, paginas):
        self._titulo = titulo
        self._autor = autor
        self._paginas = paginas

        # Revisamos que los datos sean correctos
        self.titulo = titulo
        self.autor = autor
        self.paginas = paginas

    # Propiedad titulo
    @property
    def titulo(self):
        return self._titulo

    @titulo.setter
    def titulo(self, valor):
        if not isinstance(valor, str):
            raise TypeError("El titulo debe ser un texto.")

        if valor.strip() == "":
            raise ValueError("El titulo no puede estar vacio.")

        self._titulo = valor.strip()

    # Propiedad autor
    @property
    def autor(self):
        return self._autor

    @autor.setter
    def autor(self, valor):
        if not isinstance(valor, str):
            raise TypeError("El autor debe ser un texto.")

        if valor.strip() == "":
            raise ValueError("El autor no puede estar vacio.")

        self._autor = valor.strip()

    # Propiedad paginas
    @property
    def paginas(self):
        return self._paginas

    @paginas.setter
    def paginas(self, valor):
        if isinstance(valor, bool) or not isinstance(valor, int):
            raise TypeError("Las paginas deben ser un numero entero.")

        if valor <= 0:
            raise ValueError("Las paginas deben ser mayores que cero.")

        self._paginas = valor

    # Metodo para mostrar informacion
    def mostrar_informacion(self):
        print("Titulo:", self.titulo)
        print("Autor:", self.autor)
        print("Paginas:", self.paginas)


# ============================================
# CREACION DE TRES OBJETOS INDEPENDIENTES
# ============================================

libro1 = Libro("El Principito", "Antoine de Saint-Exupery", 96)

libro2 = Libro("Don Quijote de la Mancha",
                "Miguel de Cervantes", 863)

libro3 = Libro("Cien años de soledad",
                "Gabriel Garcia Marquez", 471)


# Mostrar informacion de los libros
print("===== LIBRO 1 =====")
libro1.mostrar_informacion()

print("\n===== LIBRO 2 =====")
libro2.mostrar_informacion()

print("\n===== LIBRO 3 =====")
libro3.mostrar_informacion()
```

**2.Ejemplo de uso esperado**

```python

# Crear un objeto
libro = Libro("El principito",
               "Antoine de Saint-Exupery", 96)

# Mostrar informacion
libro.mostrar_informacion()

# Actualizacion valida
libro.paginas = 100

print("\nPaginas actualizadas:", libro.paginas)

# Actualizacion invalida
try:
    libro.paginas = -5

except ValueError as error:
    print("Error capturado:", error)

# Verifica que conserva el valor anterior
print("Paginas despues del error:", libro.paginas)
```

3.1 Creación y actualización válida, y límite de una página

```python
# Entrada: libro con una pagina
# Accion: crear un objeto
# Resultado esperado: creacion correcta

libro_prueba = Libro("Libro de prueba", "Autor de prueba", 1)

print("Creacion correcta:")
libro_prueba.mostrar_informacion()


# Entrada: 50 paginas
# Accion: actualizar paginas
# Resultado esperado: aceptar el cambio

libro_prueba.paginas = 50

print("\nActualizacion valida:")
print("Paginas:", libro_prueba.paginas)


# Entrada: 1 pagina
# Accion: actualizar paginas
# Resultado esperado: aceptar el limite minimo

libro_prueba.paginas = 1

print("\nLimite minimo aceptado:")
print("Paginas:", libro_prueba.paginas)
```

3.2 Rechazo de título, autor vacío y tipo incorrecto

```python
libro_prueba = Libro("Libro original", "Autor original", 100)


# Entrada: titulo vacio
# Accion: modificar titulo
# Resultado esperado: ValueError

try:
    libro_prueba.titulo = "   "

except ValueError as error:
    print("Titulo vacio rechazado:", error)

print("Titulo conservado:", libro_prueba.titulo)


# Entrada: autor vacio
# Accion: modificar autor
# Resultado esperado: ValueError

try:
    libro_prueba.autor = ""

except ValueError as error:
    print("\nAutor vacio rechazado:", error)

print("Autor conservado:", libro_prueba.autor)


# Entrada: titulo con tipo incorrecto
# Accion: asignar un numero al titulo
# Resultado esperado: TypeError

try:
    libro_prueba.titulo = 123

except TypeError as error:
    print("\nTipo de titulo incorrecto:", error)

print("Titulo despues del error:", libro_prueba.titulo)


# Entrada: autor con tipo incorrecto
# Accion: asignar un numero al autor
# Resultado esperado: TypeError

try:
    libro_prueba.autor = 456

except TypeError as error:
    print("\nTipo de autor incorrecto:", error)

print("Autor despues del error:", libro_prueba.autor)
```

3.3. Rechazo de páginas negativas, cero, decimales y booleanos

```python
libro_prueba = Libro("Libro de prueba", "Autor de prueba", 100)


# Entrada: paginas negativas
# Accion: actualizar paginas
# Resultado esperado: ValueError

try:
    libro_prueba.paginas = -5

except ValueError as error:
    print("Paginas negativas rechazadas:", error)

print("Valor conservado:", libro_prueba.paginas)


# Entrada: cero paginas
# Accion: actualizar paginas
# Resultado esperado: ValueError

try:
    libro_prueba.paginas = 0

except ValueError as error:
    print("\nCero paginas rechazadas:", error)

print("Valor conservado:", libro_prueba.paginas)


# Entrada: paginas decimales
# Accion: actualizar paginas
# Resultado esperado: TypeError

try:
    libro_prueba.paginas = 10.5

except TypeError as error:
    print("\nPaginas decimales rechazadas:", error)

print("Valor conservado:", libro_prueba.paginas)


# Entrada: valor booleano
# Accion: actualizar paginas
# Resultado esperado: TypeError

try:
    libro_prueba.paginas = True

except TypeError as error:
    print("\nValor booleano rechazado:", error)

print("Valor conservado:", libro_prueba.paginas)
```
